# Task 6: Lists vs. Tuples
Explain the difference between lists and tuples in Python.
- List: Mutable; its elements can be changed after creation. Lists use square brackets [].
- Tuple: Immutable; its elements cannot be changed after creation. Tuples use parentheses ().
Example:

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
```

# Task 7: Extracting a Domain from an Email
Write a function that extracts the domain name from an email address.

```python
def domain(email):
    return email.split("@")[1]

print(domain("user@domain.com"))
```

# Task 8: Checking for the Word "Dog"
Write a function that checks whether a string contains the word "dog", regardless of capitalization.

```python
def has_dog(text):
    return "dog" in text.lower()

print(has_dog("My DOG is cute"))
```

# Task 9: Counting Occurrences of "Dog"
Write a function that counts how many times "dog" appears in a string, ignoring capitalization.

```python
def count_dog(text):
    return text.lower().count("dog")

print(count_dog("dog Dog hotdog"))
```

# Task 10: Speeding Ticket Calculator

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(60, False))
print(caught_speeding(85, True))
```
