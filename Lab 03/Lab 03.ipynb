## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
import pandas as pd
import numpy as np

# 1. Load original dataset fresh
df = pd.read_csv('pakistan_used_cars.csv')

# 3(a) Missing values: Drop rows with no Brand or Model, fill others with 'Unknown'
df = df.dropna(subset=['Brand', 'Model'])
df['Registered City'] = df['Registered City'].fillna('Unknown')
df['Fuel'] = df['Fuel'].fillna('Unknown')
df['Transaction Type'] = df['Transaction Type'].fillna('Unknown')

# 3(b) Formatting: Clean city names and Price string formatting
df['Registered City'] = df['Registered City'].str.strip().str.title()
df['Price'] = df['Price'].astype(str).str.replace('Rs', '', regex=False).str.replace(',', '', regex=False).str.strip()

# 3(c) Data types: Convert Price to int64 and Year to Int64
df['Price'] = df['Price'].astype('int64')
df['Year'] = df['Year'].astype('Int64')

# 3(d) Inconsistent categories: Fix spelling mistakes in City and Model
df['Registered City'] = df['Registered City'].replace({'Sukkar': 'Sukkur'})
df['Model'] = df['Model'].replace({'Corrolla Altis': 'Corolla Altis'})

# 3(e) Duplicates: Drop exact duplicate ads
df = df.drop_duplicates()

# 3(f) Invalid values: KMs driven, fake prices, and Suzuki/Daihatsu > 5M filter
df.loc[(df['KMs Driven'] < 100) | (df['KMs Driven'] > 1_000_000) | (df['KMs Driven'].isin([12345, 123456])), 'KMs Driven'] = np.nan
df = df[~df['Price'].isin([12345678, 87654321])]
df = df[~((df['Brand'].isin(['Suzuki', 'Daihatsu'])) & (df['Price'] > 5_000_000))]

# Final check shape after Task 3
print("Task 3 Final Shape:", df.shape)
```

## Part(a):


```python
# T4a: Create Price_lakh column
df['Price_lakh'] = df['Price'] / 100_000

print("T4a Output - First 5 rows with Price_lakh:")
display(df[['Brand', 'Model', 'Price', 'Price_lakh']].head(5))
```

## Part(b):

```python
# T4b: Group by Brand to get count of cars and average price
brand_summary = df.groupby('Brand').agg(
    Total_Cars=('Model', 'count'),
    Avg_Price=('Price', 'mean')
).reset_index().sort_values(by='Total_Cars', ascending=False)

print("T4b Output - Brand-wise Summary (Top 10):")
display(brand_summary.head(10))
```

## Part(c):

```python
# T4c: Find the most expensive car brand on average or filter top brands
print("\nT4c Output - Top 5 Brands with Highest Average Price:")
top_expensive_brands = brand_summary.sort_values(by='Avg_Price', ascending=False)
display(top_expensive_brands.head(5))
```

## Part(d):

```python
# T4d: Grouping by Year and finding the average price per year
year_price_summary = df.groupby('Year')['Price'].mean().reset_index().sort_values(by='Year', ascending=False)

print("T4d Output - Average Price by Year (Top recent years):")
display(year_price_summary.head(10))
```

## Part(e):

```python
# T4e: Grouping by multiple columns (e.g., Brand and Condition) to see price trends
brand_condition_summary = df.groupby(['Brand', 'Condition'])['Price'].mean().reset_index()

print("\nT4e Output - Average Price by Brand and Condition (Top 10 rows):")
display(brand_condition_summary.head(10))
```

## Part(f):

```python
# T4f: Pivot Table - Average Price across Brand and Fuel type
pivot_df = df.pivot_table(values='Price', index='Brand', columns='Fuel', aggfunc='mean')

print("T4f Output - Pivot Table (Brand vs Fuel Average Price):")
display(pivot_df.head(10))
```

## Part(g):

```python
# T4g: Finding the most expensive car model for each Brand using idxmax
idx = df.groupby('Brand')['Price'].idxmax()
top_cars_per_brand = df.loc[idx, ['Brand', 'Model', 'Year', 'Price']].sort_values(by='Price', ascending=False)

print("\nT4g Output - Most Expensive Car per Brand:")
display(top_cars_per_brand.head(10))
```

## Part(h):

```python
# T4h: Number of ads and median price for Cash vs Installment/Leasing
t4h_summary = df.groupby('Transaction Type')['Price'].agg(
    Number_of_Ads=('count'),
    Median_Price=('median')
).reset_index()

print("T4h Output - Cash vs Installment/Leasing Analysis:")
display(t4h_summary)
```
