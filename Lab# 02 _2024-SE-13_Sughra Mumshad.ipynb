```python
np.random.seed(42)

rolls = np.random.randint(1, 7, size=10000)

print("First 20 Dice Rolls:")
print(rolls[:20])

print("Total Number of Rolls:", len(rolls))
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

## Part (a): Add 5 grace marks (maximum 100)


```python
grace_marks = np.clip(marks + 5, 0, 100)

print("T9a - Marks After Adding Grace Marks:")
print(grace_marks)
```

## Part (b): Calculate pass percentage

```python
pass_mark = 50

passed = grace_marks >= pass_mark
pass_percentage = passed.mean() * 100

print("T9b - Pass Percentage:")
print(pass_percentage, "%")
```

## Part (c): Students who passed only because of grace marks

```python
grace_passed = (marks < pass_mark) & (grace_marks >= pass_mark)

print("T9c - Students Passing Only Because of Grace Marks:")
print(np.where(grace_passed)[0])

print("Their Original Marks:")
print(marks[grace_passed])

print("Their New Marks:")
print(grace_marks[grace_passed])
```

## Part (d): Calculate z-scores

```python
z_scores = (marks - marks.mean()) / marks.std()

print("T9d - Z-Scores:")
print(z_scores)
```

## Part(a) – Create an 8×8 Canvas and Draw 7


```python

import numpy as np
import matplotlib.pyplot as plt

# Create 8x8 array
img = np.zeros((8, 8), dtype=int)

# Draw number 7
img[1, 1:7] = 255
img[2, 5] = 128
img[3, 4] = 128
img[4, 3] = 128
img[5, 2] = 128
img[6, 1] = 128

# Print array
print("Original Image:")
print(img)

# Display image
plt.figure(figsize=(5, 5))
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title("Original Number 7")
plt.xticks(range(8))
plt.yticks(range(8))
plt.grid()
plt.show()
```

## Part(b) – Flip the Image

```python

# (b) Flip image left to right

flipped_img = np.fliplr(img)

print("Flipped Image:")
print(flipped_img)

# Display flipped image
plt.figure(figsize=(5, 5))
plt.imshow(flipped_img, cmap='gray', vmin=0, vmax=255)
plt.title("Flipped Image")
plt.xticks(range(8))
plt.yticks(range(8))
plt.grid()
plt.show()
```

## Part (c): Invert the Image

```python

# (c) Invert the image

inverted_img = 255 - img

print("Inverted Image:")
print(inverted_img)

# Display inverted image
plt.figure(figsize=(5, 5))
plt.imshow(inverted_img, cmap='gray', vmin=0, vmax=255)
plt.title("Inverted Image")
plt.xticks(range(8))
plt.yticks(range(8))
plt.grid()
plt.show()
```

## Part (d): Crop the Image

```python

cropped_img = img[1:5, 3:7]

print(" Cropped Image:")
print(cropped_img)
```

## Part (e): Brighten the Image

```python

bright_img = np.clip(img.astype(np.int16) + 80, 0, 255)

print("Brightened Image:")
print(bright_img)
```

## Part (f): Image Shape, Data Type and White Pixels

```python
#  Check image properties

print(" Image Shape:", img.shape)

print("Image Data Type:", img.dtype)

white_pixels = np.sum(img == 255)
print(" Number of White Pixels:", white_pixels)
```
